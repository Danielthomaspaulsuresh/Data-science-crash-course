###  we are interseted to converts units of height from Inches to CM and weight from pounds to Kilograms(Kg)

```python
height_pred_copy = height_pred.copy()
```

```python
del height_pred_copy['Gender']
```

```python
height_pred_copy
```

```python
height_pred_copy['height_cm'] = height_pred_copy['Height']*2.54
```

```python
height_pred_copy['weight_kg']= height_pred_copy['Weight']/2.205
```

```python
height_pred_copy
```

```python
height_pred_copy.drop(['Height','Weight'],axis = 1,inplace =True)
```
